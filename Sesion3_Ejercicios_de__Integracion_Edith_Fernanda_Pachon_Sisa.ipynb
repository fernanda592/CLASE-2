# EJERCICIO **1**

```python
import pandas as pd

# Definimos el DataFrame de clientes de ejemplo
clientes = pd.DataFrame({
    "id_cliente": ["C001", "C002", "C003", "C004", "C005"],
    "nombre": ["Ana Gómez", "Luis Pérez", "María Ruiz", "Carlos Paz", "Sofía Vega"],
    "ciudad": ["Bogotá", "Medellín", "Cali", "Barranquilla", "Bogotá"],
    "segmento": ["Oro", "Plata", "Bronce", "Oro", "Plata"]
})

# Definimos el DataFrame de pedidos de ejemplo
pedidos = pd.DataFrame({
    "id_pedido": [101, 102, 103, 104],
    "cliente": ["C001", "C003", "C001", "C009"],  # C009 es un cliente inexistente para probar el merge outer
    "monto": [150.5, 200.0, 50.0, 300.0]
})

print("DataFrames 'clientes' y 'pedidos' creados correctamente.")
display(clientes)
display(pedidos)
```

# EJERCICIO **2**

```python
import pandas as pd
import numpy as np

# Aseguramos la consistencia de los números aleatorios fijando la semilla
np.random.seed(42)

# 1. Combinamos 'clientes' y 'pedidos'
consolidado = pd.merge(clientes, pedidos, left_on="id_cliente", right_on="cliente", how="inner")

# 2. Generamos los registros de ventas de ejemplo para cada mes y sucursal
consolidado_ejemplo = []
sucursales = ["Norte", "Sur"]
meses = ["Ene", "Feb", "Mar"]

for idx, row in consolidado.iterrows():
    for mes in meses:
        consolidado_ejemplo.append({
            "id_cliente": row["id_cliente"],
            "nombre": row["nombre"],
            "sucursal": np.random.choice(sucursales),
            "mes": mes,
            "ventas": row["monto"] * np.random.uniform(0.8, 1.5)
        })

consolidado = pd.DataFrame(consolidado_ejemplo)

# 3. Creamos la tabla dinámica con las ventas por sucursal y mes
t = consolidado.pivot_table(index="sucursal", columns="mes", values="ventas", aggfunc="sum")

# 4. Calculamos el porcentaje de crecimiento de Marzo respecto a Enero
t["crecimiento_%"] = ((t["Mar"] - t["Ene"]) / t["Ene"] * 100).round(1)

# 5. Ordenamos de mayor a menor crecimiento
t = t.sort_values("crecimiento_%", ascending=False)

# 6. Mostramos los resultados idénticos obtenidos anteriormente
print("Porcentaje de crecimiento por sucursal:")
print(t["crecimiento_%"])
print("\nLíder de crecimiento:", t.index[0])
```
