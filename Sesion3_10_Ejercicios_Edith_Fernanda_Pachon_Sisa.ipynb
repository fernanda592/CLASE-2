```python
# Dividir la cuenta
cuenta = 250000
propina = 25000
amigos = 3

# 1. Regresa la cuenta más la propina
def con_propina(cuenta, propina):
    return cuenta + propina

# 2. Regresa cuánto paga cada persona
def por_persona(total, personas):
    return total / personas

# 3. Úsalas con tus datos e imprime
total = con_propina(cuenta, propina)
cada_uno = por_persona(total, amigos)

print("Total con propina:", total)
print("Cada persona paga:", round(cada_uno))
```
